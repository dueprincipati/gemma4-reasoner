# Gemma 4 Multimodal Reasoner — Demo

This notebook demonstrates the core capabilities of the Gemma 4 Reasoner.

```python
from gemma4_reasoner import Gemma4Reasoner, ReasoningConfig, ModelSize, BackendType
from gemma4_reasoner.tools import ChartAnalyzer, DocumentParser, MathSolver, ScreenAnalyzer
from PIL import Image
import matplotlib.pyplot as plt

# Initialize
config = ReasoningConfig(
    model_size=ModelSize.E4B,
    backend=BackendType.OLLAMA,
    thinking=True,
)
reasoner = Gemma4Reasoner(config)
print(f"Ready! Model: {config.model_size.value}, Thinking: {config.thinking}")
```

## 1. Image Analysis

```python
img = Image.open("../data/sample_chart.png")
plt.imshow(img)
plt.axis('off')
plt.show()

result = reasoner.analyze_image(img, question="What does this chart show?")
print("Response:", result.response)
if result.has_thinking:
    print("\nThinking:", result.thinking)
```

## 2. Document Analysis

```python
docs = DocumentParser(reasoner)
result = docs.parse_table("../data/table.png")
print(result)
```

## 3. Math Problem

```python
solver = MathSolver(reasoner)
result = solver.solve("../data/math_problem.jpg")
print("Answer:", result.response)
print("\nWork:", result.thinking)
```

## 4. Screen Analysis

```python
screen = ScreenAnalyzer(reasoner)
print(screen.describe("../data/screenshot.png"))
```
